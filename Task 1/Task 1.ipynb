```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


data = pd.read_csv('Metadata_Country_API_SP.POP.TOTL_DS2_EN_csv_v2_3107.csv')


print("Dataset ki shuruat ki 5 rows:")
print(data.head())
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(x='Region', data=data, palette='viridis', order=data['Region'].value_counts().index)

plt.title('Distribution of Countries by Region', fontsize=14, fontweight='bold')
plt.xlabel('Region', fontsize=12)
plt.ylabel('Count of Countries', fontsize=12)
plt.xticks(rotation=45, ha='right') 
plt.tight_layout()


plt.savefig('region_distribution_bar_chart.png')
plt.show()
```

```python

plt.figure(figsize=(10, 6))

sns.countplot(x='IncomeGroup', data=data, palette='Set2')

plt.title('Distribution of Countries by Income Group', fontsize=14, fontweight='bold')
plt.xlabel('Income Group', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.xticks(rotation=30)
plt.tight_layout()

plt.savefig('income_distribution_histogram.png')
plt.show()
```
