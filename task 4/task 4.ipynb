```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv('twitter_training.csv', names=['ID', 'Topic', 'Sentiment', 'Text'])


print("Data Preview:")
display(df.head())


df.dropna(inplace=True)

print("\n--- Generating Sentiment Analysis Graphs ---")


plt.figure(figsize=(8, 6))
sns.countplot(x='Sentiment', data=df, palette='viridis', order=df['Sentiment'].value_counts().index)
plt.title('Overall Sentiment Distribution in Social Media Data', fontsize=14, fontweight='bold')
plt.xlabel('Sentiment Type', fontsize=12)
plt.ylabel('Count of Posts', fontsize=12)
plt.tight_layout()
plt.savefig('overall_sentiment_distribution.png') 
plt.show()


top_topics = df['Topic'].value_counts().head(5).index
df_top = df[df['Topic'].isin(top_topics)]

plt.figure(figsize=(12, 6))
sns.countplot(x='Topic', hue='Sentiment', data=df_top, palette='Set2')
plt.title('Sentiment Patterns for Top 5 Trending Topics', fontsize=14, fontweight='bold')
plt.xlabel('Topic / Brand', fontsize=12)
plt.ylabel('Count of Posts', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('sentiment_by_topic.png') 
plt.show()
```
